**1.	Difference between Git and GitHub**
Git is a version control system used to track changes in code and manage different versions of files. It works locally on your computer and allows you to save, compare, and revert changes.
GitHub is an online platform that uses Git to store code in the cloud. It allows developers to share their code, collaborate with others, and manage projects.
In simple terms, Git is the tool and GitHub is the platform where the code is hosted.



**2.	Difference between git add and git commitgit **add is used to move changes to the staging area. It prepares the files for saving but does not actually save them.
git commit is used to permanently save the changes in the repository. It creates a snapshot of the project along with a message.
Difference: git add prepares changes git commit saves changes permanently





**3.What is the staging area in Git**
The staging area is a temporary space where changes are stored before being committed. It allows you to select specific changes that you want to include in the next commit instead of saving everything at once. The staging area, also called the index, is an intermediate space between the working directory and the repository. It stores the changes that are ready to be committed.
Purpose of staging area: It allows selective commits It helps organize commits logically It avoids committing unwanted changes
Example: If you modify three files but want to commit only one, you can add only that file to the staging area and commit it.
Workflow: Working Directory contains edited files Staging Area contains selected files for commit Repository contains permanently saved versions





**4. What happens if you modify a file after committing it**
If you modify a file after committing, Git will mark it as modified. These changes are not part of the previous commit. To save them, you need to use git add and git commit again. Each commit stores a new version of the file. After a commit, if you modify a file, Git recognizes it as a new change and marks the file as modified. These changes are not automatically included in the previous commit.
To save the new changes: You must run git add again to stage the changes Then run git commit to create a new snapshot
Important concepts: Each commit is independent and represents a specific state of the project Git does not overwrite previous commits You can view differences using git diff You can check status using git status
This ensures a complete history of all changes is maintained.








**5.Difference between git push and git pull git **push is used to upload your local changes to a remote repository such as GitHub.
git pull is used to download the latest changes from the remote repository to your local system.
Difference: git push sends your changes to the server git pull brings changes from the server to your system git push is used to upload your committed changes from the local repository to a remote repository such as GitHub. It is mainly used to share your work with others. Example: git push origin main git pull is used to fetch and merge changes from the remote repository into your local repository. It ensures your code is up to date with the latest version.
Example: git pull origin main
Key points: git push sends data from local to remote git pull brings data from remote to local git pull is actually a combination of git fetch and git merge Conflicts may occur during pull if multiple users modify the same code 6.What is Version Control Centralized vs Distributed VCS
Version Control is a system that tracks changes in files over time. It helps manage different versions and allows multiple people to work on the same project.
Centralized Version Control System: There is one central server where all files are stored. Users connect to this server to access and update files. It requires internet and if the server fails, data may be lost.
Distributed Version Control System: Each user has a complete copy of the repository. Work can be done offline and changes can be shared later. It is faster and more secure because data is stored in multiple places.
Example: Centralized VCS example is SVN Distributed VCS example is Git

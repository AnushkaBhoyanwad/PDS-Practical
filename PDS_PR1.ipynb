## 1. What are sequence data types in Python? Name them and explain how indexing and slicing work across different sequence types?

Sequence data types are data structures that store multiple elements in an ordered manner. Each element is assigned a position (index) which allows accessing elements individually.

1. **String (str):** A sequence of characters.
2. **List (list):** An ordered, mutable collection of elements.
3. **Tuple (tuple):** An ordered, immutable collection of elements.
4. **Range (range):** An immutable sequence of numbers, often used in loops.

### Indexing
Indexing means accessing individual elements of a sequence using their position (index). In Python, indexing is zero-based, meaning the first element is at index 0.

```python
text = "Python"
print(text[0])   # Output: P
print(text[3])   # Output: h
print(text[-1])  # Output: n (negative indexing accesses from the end)
```

### Slicing
Slicing extracts a portion (a subsequence) from a sequence. It uses the syntax `sequence[start : end : step]`:
- **start:** The starting index of the slice (inclusive). If omitted, defaults to the beginning.
- **end:** The ending index of the slice (exclusive). If omitted, defaults to the end.
- **step:** The step size between elements. If omitted, defaults to 1.

```python
my_list = [10, 20, 30, 40, 50, 60]
print(my_list[1:5:2])  # Output: [20, 40]
```

## 2. Difference between List and Tuple in Python

| Feature | List | Tuple |
|---------|------|-------|
| Mutability | Mutable (can change) | Immutable (cannot change) |
| Syntax | `[ ]` | `( )` |
| Performance | Slightly slower | Faster |
| Memory | Uses more memory | Uses less memory |

### Why prefer tuples in real-world applications?
- **Data should not change:** Tuples ensure data integrity for constant values.
- **Better performance and memory efficiency:** Tuples are more memory-efficient and slightly faster.
- **Used as dictionary keys:** Immutable tuples can be dictionary keys; mutable lists cannot.
- **Safer for data integrity:** Contents cannot be accidentally altered once created.

## 3. How Python arrays are different from lists

Python lists are general-purpose containers that can store items of different data types, while arrays (from the `array` module) are designed for efficient storage of homogeneous (same data type) numeric data.

| Feature | List | Array |
|---------|------|-------|
| Data types | Can store mixed data types | Stores elements of the same data type |
| Performance | Slower for numeric operations | Faster for numeric operations |
| Memory | Uses more memory | More memory-efficient and compact |
| Flexibility | Highly flexible, dynamic size | Less flexible, fixed data type |
| Built-in | Built-in data type | Requires importing the `array` module |

### When should you use arrays?
- Storing large numeric datasets
- When memory efficiency is critical
- Performing scientific or numerical computations (NumPy preferred for serious work)

## 4. How dictionaries work internally in Python

A dictionary in Python is an unordered collection of data values, used to store key:value pairs.

### Internal Mechanism: Hash Tables
Internally, Python dictionaries are implemented using a **hash table**.

**How it works:**
1. **Hashing the Key:** Python applies a `hash()` function to the key, converting it to an integer (hash value).
2. **Mapping to Memory Index:** The hash value determines a specific memory location ("bucket") where the key-value pair is stored.
3. **Storing the Pair:** The value is stored along with its key at that memory index.

### Why dictionary lookup is fast
Dictionary access is typically O(1) (constant time) on average.

| Data Structure | Search Time (Average) |
|----------------|----------------------|
| List | O(n) |
| Tuple | O(n) |
| Dictionary | O(1) |

**Common uses:** Database lookups, caching, data indexing, counting frequencies.

## 5. Difference between Sets and Lists in Python

| Feature | List | Set |
|---------|------|-----|
| Order | Ordered (maintains insertion order) | Unordered |
| Duplicates | Allowed | Not allowed (stores only unique elements) |
| Indexing/Slicing | Supported | Not supported |
| Mutability | Mutable | Mutable (but elements must be hashable) |
| Syntax | `[ ]` | `{ }` or `set()` |
| Performance | Slower for membership checking | Faster for membership checking |

### How sets handle duplicates
1. Python hashes the element to determine where it would be stored internally.
2. If an element with the same hash and value already exists, the new element is ignored.

### Practical uses of sets
- **Removing duplicates** from data
- **Fast membership testing** (average O(1))
- **Mathematical set operations:** `union()`, `intersection()`, `difference()`, `symmetric_difference()`
- **Checking for uniqueness**

````python
print("LIST MAX & MIN")
numbers = [12, 45, 7, 23, 56]
print("List:", numbers)
print("Maximum value:", max(numbers))
print("Minimum value:", min(numbers))
````

````python
print("TUPLE OPERATIONS")
students = ("Sarang", "Amit", "Rohit", "Neha", "Priya")
print("Student Names:")
for name in students:
    print(name)
````

````python
print("DICTIONARY OPERATIONS")
student_dict = {
    19: "Sarang",
    22: "Amit",
    33: "Rohit",
    44: "Neha"
}
print("Dictionary:", student_dict)
print("Roll Numbers:")
for key in student_dict.keys():
    print(key)
````

````python
print("SET OPERATIONS")
list_numbers = [1, 2, 3, 4, 2, 3, 5, 6, 1]
print("Original List:", list_numbers)
unique_set = set(list_numbers)
print("Set (Duplicates removed):", unique_set)
````

````python
print("VOWEL COUNT")
string = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0
for char in string:
    if char in vowels:
        count += 1
print("Number of vowels:", count)
````

````python
print("LIST INSERTION & DELETION")
num_list = [10, 20, 30, 40, 50]
print("Original List:", num_list)
num_list.insert(2, 25)
print("After insertion:", num_list)
num_list.remove(30)
print("After deletion:", num_list)
````

````python
print("TYPE CONVERSION")
list1 = [1, 2, 3, 4]
tuple1 = tuple(list1)
print("List to Tuple:", tuple1)

tuple2 = (5, 6, 7, 8)
list2 = list(tuple2)
print("Tuple to List:", list2)
````

````python
print("Range Example")
print("Numbers using range:")
for i in range(1, 6):
    print(i)
````
